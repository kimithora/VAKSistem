```python
import json
import os

# Lokasi file JSON
json_path = r"C:\projek\penelitian2\Kodinga\WEB\data\tokesiswa.json"

# Lokasi file SQL hasil konversi
sql_path = r"C:\projek\penelitian2\Kodinga\WEB\data\akun_siswa.sql"

# Baca JSON
with open(json_path, "r", encoding="utf-8") as file:
    data = json.load(file)

# Buat SQL
values = []

for item in data:
    token = str(item["Token"]).replace("'", "''")
    inisial = str(item["Inisial"]).replace("'", "''")
    role = str(item["Role"]).replace("'", "''")

    values.append(
        f"('{token}', '{inisial}', '{role}')"
    )

sql = "INSERT INTO akun_siswa (Token, Inisial, Role) VALUES\n"
sql += ",\n".join(values)
sql += ";"

# Simpan SQL
with open(sql_path, "w", encoding="utf-8") as file:
    file.write(sql)

print("Berhasil!")
print(f"Jumlah data: {len(data)}")
print(f"File SQL: {sql_path}")
```

```python
import pandas as pd

xlsx_path = r"C:\projek\penelitian2\Kodinga\WEB\data\SMAN1.xlsx"
sql_path = r"C:\projek\penelitian2\Kodinga\WEB\data\jawaban_siswa.sql"

# Baca Excel
df = pd.read_excel(xlsx_path)

# Ambil hanya Inisial + Q1 sampai Q60
kolom = ["Inisial"] + [f"Q{i}" for i in range(1, 61)]
df = df[kolom]

# Buat SQL
values = []

for _, row in df.iterrows():
    inisial = str(row["Inisial"]).strip().replace("'", "''")

    jawaban = []
    for i in range(1, 61):
        nilai = row[f"Q{i}"]

        if pd.isna(nilai):
            jawaban.append("NULL")
        else:
            jawaban.append(str(int(nilai)))

    values.append(
        f"(NULL, '{inisial}', {', '.join(jawaban)})"
    )

sql = """
INSERT INTO jawaban_siswa
(Token, Inisial, Q1, Q2, Q3, Q4, Q5, Q6, Q7, Q8, Q9, Q10,
 Q11, Q12, Q13, Q14, Q15, Q16, Q17, Q18, Q19, Q20,
 Q21, Q22, Q23, Q24, Q25, Q26, Q27, Q28, Q29, Q30,
 Q31, Q32, Q33, Q34, Q35, Q36, Q37, Q38, Q39, Q40,
 Q41, Q42, Q43, Q44, Q45, Q46, Q47, Q48, Q49, Q50,
 Q51, Q52, Q53, Q54, Q55, Q56, Q57, Q58, Q59, Q60)
VALUES
""" + ",\n".join(values) + ";"

with open(sql_path, "w", encoding="utf-8") as file:
    file.write(sql)

print("Berhasil!")
print("Jumlah data:", len(df))
print("File SQL:", sql_path)
```

```python
# Urutkan kolom
data = data[
    ["Token", "Inisial"] + [f"Q{i}" for i in range(1, 61)]
]

# Buat daftar nama kolom SQL
kolom_sql = ", ".join(data.columns)

# Buat isi VALUES
values = []

for _, row in data.iterrows():
    token = str(row["Token"]).replace("'", "''")
    inisial = str(row["Inisial"]).replace("'", "''")

    jawaban_q = []

    for i in range(1, 61):
        nilai = row[f"Q{i}"]

        if pd.isna(nilai):
            jawaban_q.append("NULL")
        else:
            jawaban_q.append(str(int(nilai)))

    values.append(
        f"('{token}', '{inisial}', {', '.join(jawaban_q)})"
    )

# Gabungkan menjadi SQL
sql = f"""INSERT INTO jawaban_siswa ({kolom_sql})
VALUES
{",\n".join(values)};"""

# Simpan
with open(sql_path, "w", encoding="utf-8") as file:
    file.write(sql)

print("SQL berhasil dibuat!")
print("Jumlah data:", len(data))
print("File:", sql_path)
```
